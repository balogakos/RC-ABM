```python
# 1. Define the store categories
store_types = [
    'Cafe', 'Convenience Store', 'Entertainment', 'Foodstore',
    'Personal Service', 'Professional Services', 'Restaurant', 'Retail'
]

# 2. Convert to binary and multiply by the Utility_Score
# This creates new 'Score_' columns for each type
for store in store_types:
    binary_col = (final_results[store] > 0).astype(int)
    final_results[f'Score_{store}'] = binary_col * final_results['Utility_Score']

# 3. Create a dictionary to hold the top 10 for each type
top_10_by_store = {}

for store in store_types:
    score_col = f'Score_{store}'
    # Filter for centres that actually have that store type, then sort
    top_df = (
        final_results[final_results[score_col] > 0]
        .sort_values(score_col, ascending=False)
        .head(10)
    )
    # Keep only the relevant columns for display
    top_10_by_store[store] = top_df[['RC_ID', 'Utility_Score', store]]

# 4. Display the results nicely in your notebook
import IPython.display as display

for store, df in top_10_by_store.items():
    print(f"\n{'='*20} TOP 10: {store.upper()} {'='*20}")
    if not df.empty:
        display.display(df)
    else:
        print("No matches found for this category.")
```
