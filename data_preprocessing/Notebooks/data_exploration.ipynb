```python
import pandas as pd
import os

# Set the path to the testing directory
test_data_dir = r'C:\Users\sgabalog\Documents\P3\Model\Utility\testing'

# List all parquet files in the folder
parquet_files = [f for f in os.listdir(test_data_dir) if f.endswith('.parquet')]

print(f"--- Converting parquet files to CSV in {test_data_dir} ---")

for file_name in parquet_files:
    # Full path to the input parquet
    input_path = os.path.join(test_data_dir, file_name)
    
    # Define output filename (replacing .parquet with .csv)
    output_name = file_name.replace('.parquet', '.csv')
    output_path = os.path.join(test_data_dir, output_name)
    
    # Read the parquet and save as CSV
    try:
        df = pd.read_parquet(input_path)
        df.to_csv(output_path, index=True)
        print(f"  Successfully saved: {output_name} (Rows: {len(df)})")
    except Exception as e:
        print(f"  Error converting {file_name}: {e}")

print("\n--- All conversions complete ---")
```
